# Symmetries

Define and use symmetry operations with ``SymmetryGroup`` to reduce computation. 

Example: complex conjugation symmetry ``G*(ν) = G(-ν)``.

 A one-arg symmetry mapping (frequency -> -frequency), with conjugation:

```julia
using MatsubaraFunctions

conj_sym = Symmetry{1}() do args
    ω = args[1]                   # MeshPoint
    return ( -ω, ), Operation{ComplexF64}(sgn=false, con=true)
end
```

```julia
T = 0.1     # temperature
N = 32      # number of positive frequencies

mf_fermi = MatsubaraMesh(T, N, Fermion);
```

Build a ``MeshFunction`` and a group:

```julia
G = MeshFunction(mf_fermi; data_t=ComplexF64);
SG = SymmetryGroup([conj_sym], G);
```

Provide an initializer for irreducible elements:

```julia
init = InitFunction{1, ComplexF64}(
    x -> begin
        ω = x[1]                      # MeshPoint
        1.0 / (im * value(ω) + 0.5)
    end
    )
```

Fill ``G`` over irreducible set, propagate via symmetries:

```julia
SG(G, init)
```

We can query equivalence classes or irreducible indices:

```julia
classes(SG, length(SG)) 
```

```julia
irreducible(SG, length(SG))
```
