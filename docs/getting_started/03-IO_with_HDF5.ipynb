# I/O with HDF5

```julia
using MatsubaraFunctions
using HDF5

T = 0.1     # temperature
N = 32      # number of positive frequencies

mf_fermi = MatsubaraMesh(T, N, Fermion)

Gω = MeshFunction(mf_fermi; data_t=ComplexF64);
```

Save meshes and MeshFunction objects:

```julia
h5open("matsubara_fermi.h5", "w") do file
    save!(file, "mesh/fermi", mf_fermi)
    save!(file, "func/Gω", Gω)
end
```

Load meshes and MeshFunction objects:

```julia
mf_fermi2 = h5open("matsubara_fermi.h5", "r") do file
    load_mesh(file, "mesh/fermi")
end;
```

```julia
Gω2 = h5open("matsubara_fermi.h5", "r") do file
    load_mesh_function(file, "func/Gω")
end;
```
